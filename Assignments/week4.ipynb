```python
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/wujh03/MSSP6070/main/data/parsed.csv')
df.head()
```

```python
#a. Find the 95th percentile of earthquake magnitude in Japan using magType of 'mb'
japan_mb_eqs = df[(df['parsed_place'] == 'Japan') & (df['magType'] == 'mb')]

japan_95th_percentile = japan_mb_eqs['mag'].quantile(0.95)

print(f"The 95th percentile of the magnitude of an earthquake in Japan with a magType of 'mb' is: {japan_95th_percentile}")
```

```python
#b. Find the percentage of earthquakes in Indonesia that were coupled with tsunamis.
indonesia_eqs = df[df['parsed_place'] == 'Indonesia']

tsunami_percentage = indonesia_eqs['tsunami'].mean() * 100

print(f"Among the earthquakes that occurred in Indonesia, the percentage accompanied by tsunamis was: {tsunami_percentage:.2f}%")
```

```python
#c. Get summary statistics for earthquakes in Nevada
nevada_eqs = df[df['parsed_place'] == 'Nevada']

nevada_summary = nevada_eqs.describe()

display(nevada_summary)
```

```python
#d. Add a column to the dataframe indicating whether or not the earthquake happened in a country or US state that is on the Ring of Fire.

ring_of_fire_places = [
    'Bolivia', 'Chile', 'Ecuador', 'Peru', 'Costa Rica', 'Guatemala',
    'Japan', 'Philippines', 'Indonesia', 'New Zealand',
    'Canada', 'Fiji', 'Alaska', 'Washington', 'California',
    'Russia', 'Taiwan', 'Tonga', 'Kermadec Islands'
]

def is_ring_of_fire(place):
    if not isinstance(place, str):
        return False

    if place == 'New Mexico':
        return False

    if place == 'Mexico':
        return True

    if 'Antarctic' in place:
        return True

    if any(loc in place for loc in ring_of_fire_places):
        return True

    return False

df['ring_of_fire'] = df['parsed_place'].apply(is_ring_of_fire)

display(df[['parsed_place', 'ring_of_fire']].head(15))
```

```python
#e. Calculate the number of earthquakes in the Ring of Fire locations and the number outside of them
ring_of_fire_counts = df['ring_of_fire'].value_counts()

print("Earthquakes count inside vs outside the Ring of Fire:")
print(ring_of_fire_counts)

in_ring = ring_of_fire_counts.get(True, 0)
out_ring = ring_of_fire_counts.get(False, 0)

print(f"\nNumber of earthquakes inside the Ring of Fire: {in_ring}")
print(f"Number of earthquakes outside the Ring of Fire: {out_ring}")
```
