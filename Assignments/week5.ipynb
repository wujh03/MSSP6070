```python
import pandas as pd
```

```python
aapl = pd.read_csv('data/aapl.csv')
amzn = pd.read_csv('data/amzn.csv')
fb = pd.read_csv('data/fb.csv')
nflx = pd.read_csv('data/nflx.csv')
goog = pd.read_csv('data/goog.csv')
```

```python
aapl.head()
```

```python
aapl['ticker'] = 'AAPL'
amzn['ticker'] = 'AMZN'
fb['ticker'] = 'FB'
nflx['ticker'] = 'NFLX'
goog['ticker'] = 'GOOG'
```

```python
faanf = pd.concat(
    [fb, aapl, amzn, nflx, goog],
    ignore_index=True
)

faanf.head()
```

```python
faanf.to_csv('data/faanf.csv', index=False)
```

```python
faanf['date'] = pd.to_datetime(faanf['date'])
faanf['volume'] = faanf['volume'].astype(int)

faanf.dtypes
```

```python
faanf = faanf.sort_values(
    by=['date', 'ticker']
).reset_index(drop=True)

faanf.head(10)
```

```python
top_7_volume = faanf.nlargest(7, 'volume')
top_7_volume
```

```python
faanf_melted = faanf.melt(
    id_vars=['date', 'ticker']
)

faanf_melted.head(20)
```

```python
faanf.to_csv('data/faanf.csv', index=False)
```
