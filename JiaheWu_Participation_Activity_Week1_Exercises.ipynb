## Exercise 4: Generate the data by running this cell

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
manual_mean = sum(salaries) / len(salaries)
print("Manual Mean:", manual_mean)
import statistics
stats_mean = statistics.mean(salaries)
print("Statistics Mean:",stats_mean)
print("Matches?", manual_mean==stats_mean)
```

### median

```python
sorted_salaries = sorted(salaries)
n = len(sorted_salaries)

if n % 2 == 1:
    manual_median = sorted_salaries[n // 2]
    manual_median = (sorted_salaries[(n // 2) - 1] + sorted_salaries[n // 2]) / 2
print("Manual Median:", manual_median)


import statistics

stats_median = statistics.median(salaries)
print("Statistics Median:", stats_median)

print("Matches?", manual_median == stats_median)
```

### mode

```python
counts = {}
for num in salaries:
    counts[num] = counts.get(num, 0) + 1

max_count = max(counts.values())

manual_modes = [num for num, count in counts.items() if count == max_count]

manual_mode = manual_modes[0] if len(manual_modes) == 1 else manual_modes
print("Manual Mode:", manual_mode)


import statistics

stats_mode = statistics.mode(salaries)
print("Statistics Mode:", stats_mode)

if isinstance(manual_mode, list):
    print("Mode Matches?", stats_mode in manual_mode)
else:
    print("Mode Matches?", manual_mode == stats_mode)
```

### sample variance
Remember to use Bessel's correction.

```python
mean_val = sum(salaries) / len(salaries)

manual_variance = sum((x - mean_val) ** 2 for x in salaries) / (len(salaries) - 1)
print("Manual Sample Variance:", manual_variance)


import statistics

stats_variance = statistics.variance(salaries)
print("Statistics Sample Variance:", stats_variance)


print("Matches?", manual_variance== stats_variance)
```

### sample standard deviation
Remember to use Bessel's correction.

```python
import math
import statistics


mean_val = sum(salaries) / len(salaries)

sample_variance = sum((x - mean_val) ** 2 for x in salaries) / (len(salaries) - 1)

manual_stdev = math.sqrt(sample_variance)
print("Manual Sample StDev:", manual_stdev)



stats_stdev = statistics.stdev(salaries)
print("Statistics Sample StDev:", stats_stdev)


print("Matches?", math.isclose(manual_stdev, stats_stdev))
```

## Exercise 6: Calculating more statistics
### range

```python
manual_range = max(salaries) - min(salaries)
print("Manual Range:", manual_range)

sorted_salaries = sorted(salaries)
sorted_range = sorted_salaries[-1] - sorted_salaries[0]
print("Sorted Range:", sorted_range)


print("Matches?", manual_range == sorted_range)
```

### coefficient of variation

```python
import statistics

mean_val = sum(salaries) / len(salaries)

sample_variance = sum((x - mean_val) ** 2 for x in salaries) / (len(salaries) - 1)
sample_stdev = sample_variance ** 0.5

manual_cv = sample_stdev / mean_val
print("Manual CV:", manual_cv)
print(f"Manual CV (Percentage): {manual_cv * 100:.2f}%")

stats_cv = statistics.stdev(salaries) / statistics.mean(salaries)
print("Statistics CV:", stats_cv)


import math
print("Matches?", math.isclose(manual_cv, stats_cv))
```

### interquartile range

```python
import statistics
import math

sorted_salaries = sorted(salaries)
n = len(sorted_salaries)

q1_index = 0.25 * (n - 1)
q3_index = 0.75 * (n - 1)

i1, f1 = int(q1_index), q1_index % 1
i3, f3 = int(q3_index), q3_index % 1

manual_q1 = sorted_salaries[i1] + f1 * (sorted_salaries[i1 + 1] - sorted_salaries[i1])
manual_q3 = sorted_salaries[i3] + f3 * (sorted_salaries[i3 + 1] - sorted_salaries[i3])

manual_iqr = manual_q3 - manual_q1
print("Manual Q1:", manual_q1)
print("Manual Q3:", manual_q3)
print("Manual IQR:", manual_iqr)


q1, q2, q3 = statistics.quantiles(salaries, n=4)
stats_iqr = q3 - q1

print("Statistics Q1:", q1)
print("Statistics Q3:", q3)
print("Statistics IQR:", stats_iqr)


print("Matches?", math.isclose(manual_iqr, stats_iqr))
```

### quartile coefficent of dispersion

```python
import statistics
import math

sorted_salaries = sorted(salaries)
n = len(sorted_salaries)

q1_idx = 0.25 * (n - 1)
q3_idx = 0.75 * (n - 1)

i1, f1 = int(q1_idx), q1_idx % 1
i3, f3 = int(q3_idx), q3_idx % 1

manual_q1 = sorted_salaries[i1] + f1 * (sorted_salaries[i1 + 1] - sorted_salaries[i1])
manual_q3 = sorted_salaries[i3] + f3 * (sorted_salaries[i3 + 1] - sorted_salaries[i3])

manual_qcod = (manual_q3 - manual_q1) / (manual_q3 + manual_q1)
print("Manual Quartile Coeff of Dispersion:", manual_qcod)


q1, q2, q3 = statistics.quantiles(salaries, n=4)

stats_qcod = (q3 - q1) / (q3 + q1)
print("Statistics Quartile Coeff of Dispersion:", stats_qcod)

print("Matches?", math.isclose(manual_qcod, stats_qcod))
```

## Exercise 7: Scaling data
### min-max scaling

```python
min_val = min(salaries)
max_val = max(salaries)

scaled_salaries = [(x - min_val) / (max_val - min_val) for x in salaries]

print("Scaled Salaries (First 5):", scaled_salaries[:5])

print("New Min:", min(scaled_salaries))
print("New Max:", max(scaled_salaries))
```

### standardizing

```python
import statistics
import math

mean_val = sum(salaries) / len(salaries)
stdev_val = (sum((x - mean_val) ** 2 for x in salaries) / (len(salaries) - 1)) ** 0.5

standardized_salaries = [(x - mean_val) / stdev_val for x in salaries]

print("Standardized Salaries (First 5):", standardized_salaries[:5])


new_mean = statistics.mean(standardized_salaries)
new_stdev = statistics.stdev(standardized_salaries)

print("New Mean (should be 0):", new_mean)
print("New StDev (should be 1):", new_stdev)

print("Mean is ~0?", math.isclose(new_mean, 0, abs_tol=1e-9))
print("StDev is 1?", math.isclose(new_stdev, 1))
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
import math
import random
import statistics

random.seed(1)
salaries_2 = [round(random.random() * 1000000, -3) for _ in range(len(salaries))]

mean_1 = sum(salaries) / len(salaries)
mean_2 = sum(salaries_2) / len(salaries_2)

manual_cov = sum((x - mean_1) * (y - mean_2) for x, y in zip(salaries, salaries_2)) / (
    len(salaries) - 1
)
print("Manual Covariance:", manual_cov)


stats_cov = statistics.covariance(salaries, salaries_2)
print("Statistics Covariance:", stats_cov)


print("Matches?", math.isclose(manual_cov, stats_cov))
```

### Pearson correlation coefficient ($\rho$)

```python
import math
import statistics

mean_1 = sum(salaries) / len(salaries)
mean_2 = sum(salaries_2) / len(salaries_2)

cov = sum((x - mean_1) * (y - mean_2) for x, y in zip(salaries, salaries_2)) / (
    len(salaries) - 1
)

stdev_1 = (sum((x - mean_1) ** 2 for x in salaries) / (len(salaries) - 1)) ** 0.5
stdev_2 = (
    sum((y - mean_2) ** 2 for y in salaries_2) / (len(salaries_2) - 1)
) ** 0.5

manual_corr = cov / (stdev_1 * stdev_2)
print("Manual Pearson Correlation:", manual_corr)

stats_corr = statistics.correlation(salaries, salaries_2)
print("Statistics Pearson Correlation:", stats_corr)

print("Matches?", math.isclose(manual_corr, stats_corr))
```
