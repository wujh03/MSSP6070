```python
letters=['a',  'b',  'c', 'd']
val_3b=letters[int(int('3'*2)//11)] # 3'*2 gets the string '33', int('33') is converted to the integer 33, 33 // 11 divides the result into 3, and letters[3] takes the element at index 3 as 'd'.
print("3b",val_3b)

val_3c=letters[-1] # Negative index -1 indicates accessing the last element in the list, so the result is 'd'.
print("3c",val_3c)

val_3d=letters[:2] # Slicing syntax [:2] extracts sublists from index 0 to before index 2 (excluding index 2), so the result is ['a', 'b']
print("3d",val_3d)
```

```python
def format_list(items):
    if not items:
        return ""
    if len(items) == 1:
        return str(items[0])
    return ", ".join(items[:-1]) + ", and " + items[-1]


spam = ["apples", "bananas", "tofu", "cats"]
result = format_list(spam)
print(result)
```
