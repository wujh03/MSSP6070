**a. When does the code in a function execute: when the function is defined or called?**
The code inside a function executes only when the function is called, not when it is initially defined.

**b. How many global scopes are there in a Python program? How many local scopes?**
There is only one global scope created when a Python program begins execution. But a new local scope is created every time a function is called, so there can be an unlimited number of local scopes throughout the program's lifecycle.

**c. What happens to variables in a local scope when the function call returns?**
When a function call returns, its local scope is destroyed. All variables contained within that local scope are forgotten and their memory is freed.

**d. How can you prevent a program from crashing when it gets an error?**
By using exception handling. Place the potentially error-causing code inside a try block and define the fallback behavior in an except block.

```python
def collatz(number):
    if number % 2 == 0:
        result = number // 2
    else:
        result = 3 * number + 1

    print(result, end=' ')
    return result

try:
    user_input = int(input("Enter number:\n"))
    print(user_input, end=' ')

    while user_input != 1:
        user_input = collatz(user_input)

except ValueError:
    print("\nError: You must enter an integer.")
```
