```python
import pandas as pd

url = "https://raw.githubusercontent.com/wujh03/MSSP6070/main/data/Student%20Performance%20Data.csv"

df = pd.read_csv(url)

display(df.head())
print(df.columns.tolist())
```

```python
# Filter students whose math score is above average
filter_students = lambda data: data[
    data["math score"] > data["math score"].mean()
].copy()

# Apply the lambda function
new_df = filter_students(df)

# Display the new DataFrame
display(new_df.head())

print("Original shape:", df.shape)
print("New shape:", new_df.shape)
```
