```python
import pandas as pd

df = pd.read_csv(
    '/content/drive/MyDrive/MSSP6070/Modules/Week7 (1)/InternetSales.csv',
    encoding='latin1'
)

df.head()
```

```python
import matplotlib.pyplot as plt

df['OrderDate'] = pd.to_datetime(df['OrderDate'])
df['Year'] = df['OrderDate'].dt.year

yearly_sales = df.groupby('Year')['SalesAmount'].sum()

fig, ax = plt.subplots(figsize=(9, 5))
yearly_sales.plot(kind='bar', ax=ax)

ax.set_title('Yearly Internet Sales')
ax.set_xlabel('Year')
ax.set_ylabel('Total Sales ($)')

plt.xticks(rotation=0)
plt.tight_layout()

fig.savefig('Yearly_Internet_Sales.pdf')
plt.show()
```
